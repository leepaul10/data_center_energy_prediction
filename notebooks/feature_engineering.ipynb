```python
#Filling Missing meter readings as this corresponds to prediction
df['meter_reading']=(df.groupby('building_id')['meter_reading'].transform(lambda x: x.interpolate(method='linear')))

df['meter_reading']=(df.groupby('building_id')['meter_reading'].transform(lambda x: x.ffill().bfill()))


#Correct upto this much
```

```python
#Even after Interpolating these 6 missing values cannot be filled so the only solution is to drop them
df = df.dropna(subset=['meter_reading']).reset_index(drop=True)
```

```python
### There are two object columns timestamp and primary_use we need to convert it into meaningful numbers
# Here we need to use Encoding 
# We are using label encoding as 
```

```python
df['hour'] = df['timestamp'].dt.hour
df['day'] = df['timestamp'].dt.day
df['weekday'] = df['timestamp'].dt.weekday
df['month'] = df['timestamp'].dt.month
df['day'].isin([5,6]).astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder
encode=LabelEncoder()
df['primary_use']=encode.fit_transform(df['primary_use'])
```

```python
df['primary_use'].unique()
```

```python
df
```

```python
# Implementing LAG Features
df['lag_1h']  = df.groupby('building_id')['meter_reading'].shift(1)
df['lag_24h'] = df.groupby('building_id')['meter_reading'].shift(24)
df['lag_48h'] = df.groupby('building_id')['meter_reading'].shift(48)
df['y_1h']  = df.groupby('building_id')['meter_reading'].shift(-1)
df['y_24h'] = df.groupby('building_id')['meter_reading'].shift(-24)
df['y_48h'] = df.groupby('building_id')['meter_reading'].shift(-48)
df=df.drop(columns=['timestamp'])
```

```python
df = df.dropna(subset=['lag_1h','lag_24h','lag_48h','y_1h','y_24h','y_48h'])
```

```python
df
```

```python
df.groupby('building_id')[['meter_reading','lag_1h','lag_24h','lag_48h']].agg(['min','max','mean'])
```

```python
#Here we Can see some Buildings are useless and repeats the reading
useless_buildings=(df.groupby('building_id')['meter_reading'].std().reset_index().query('meter_reading == 0'))
useless_buildings
```

```python
# now we have to drop them 

df=df[~df['building_id'].isin(useless_buildings['building_id'])]
```

```python
df=df.reset_index(drop=True)
```

```python
# Finalcheck before exporting
df.isna().sum()
```
