```python
building_keep=dfmerge['building_id'].unique()[:16]
```

```python
energy_small=energy[(energy['building_id'].isin(building_keep))& (energy['timestamp']<'2016-04-01')]
```

```python
weather_small=weather[(weather['site_id'].isin(building_keep))& (weather['timestamp']<'2016-04-01')]
```

```python
# df_optimized=df.groupby('building_id',group_keys=False).apply(lambda x: x.sample(frac=0.1,random_state=42))
df_optimized=dfmerge.sample(n=65000,random_state=42)
```

```python
df_optimized
```

```python
#Choosing the necessary columnd from weather_small to merge and produce the final dataset

weather_columns=['site_id','timestamp','air_temperature','cloud_coverage','dew_temperature','precip_depth_1_hr','sea_level_pressure','wind_direction','wind_speed']


df=df_optimized.merge(weather_small[weather_columns],on=['site_id','timestamp'],how='left')

```

```python
df
```

```python
df=df.sort_values(by=['building_id','timestamp']).reset_index(drop=True)
df
```

```python
#Checking for missing values
df.isna().sum()
```

```python
#Filling the Missing Values for year_built
df['year_built']=df['year_built'].fillna(df['year_built'].median())
df.isna().sum()
```

```python
df['air_temperature']=df['air_temperature'].fillna(df['air_temperature'].median())
```

```python
df['cloud_coverage']=df['cloud_coverage'].fillna(df['cloud_coverage'].median())
```

```python
df['dew_temperature']=df['dew_temperature'].fillna(df['dew_temperature'].median())
```

```python
df['precip_depth_1_hr']=df['precip_depth_1_hr'].fillna(0)
```

```python
df['sea_level_pressure']=df['sea_level_pressure'].fillna(df['sea_level_pressure'].median())
```

```python
df['wind_direction']=df['wind_direction'].fillna(df['wind_direction'].median())
```

```python
df['wind_speed']=df['wind_speed'].fillna(df['wind_speed'].median())
```

```python
df.isna().sum()
```

```python
##Floor count is having large number of errors or missing values so drop
#drop floor

df=df.drop(columns=['floor_count'])
```
